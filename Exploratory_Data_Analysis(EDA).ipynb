## Introduction

In this task, the Titanic dataset is analyzed using Python libraries such as Pandas, Matplotlib, and Seaborn. The analysis focuses on passenger demographics, survival patterns, passenger class, age, fare, missing values, relationships between variables, and potential outliers.

Various statistical methods and visualizations such as count plots, histograms, boxplots, scatterplots, heatmaps, and pairplots are used to extract meaningful insights from the data.

## EDA Methodology

The following steps were performed during the analysis:

1. Loaded the Titanic dataset using Seaborn.
2. Inspected the dataset structure, shape, columns, and data types.
3. Checked descriptive statistics using `describe()`.
4. Identified missing values in the dataset.
5. Analyzed categorical variables such as gender, survival status, passenger class, and embarkation port.
6. Used visualizations to study distributions and relationships between variables.
7. Analyzed age and fare distributions using histograms and boxplots.
8. Examined relationships using scatterplots, correlation heatmaps, and pairplots.
9. Calculated survival rates based on gender, passenger class, and age groups.
10. Performed multivariate analysis to understand the combined effect of gender, passenger class, and survival.
11. Summarized the major findings and insights obtained from the analysis.

## Dataset Overview

The Titanic dataset contains information about passengers who travelled on the Titanic.

The dataset includes demographic, passenger class, family, fare, and survival-related information. It contains both numerical and categorical variables, making it suitable for exploratory data analysis.

### Important Variables

- **Survived:** Indicates whether the passenger survived (0 = No, 1 = Yes).
- **Pclass:** Passenger class (1 = First, 2 = Second, 3 = Third).
- **Gender:** Passenger gender.
- **Age:** Age of the passenger.
- **SibSp:** Number of siblings or spouses aboard.
- **Parch:** Number of parents or children aboard.
- **Fare:** Passenger ticket fare.
- **Embarked:** Port from which the passenger embarked.

The analysis uses statistical summaries and different visualizations to understand patterns and relationships among these variables.

## Statistical Analysis

Statistical analysis was performed to understand the basic characteristics of the Titanic dataset.

The following methods were used:

- `df.shape` to identify the number of records and variables.
- `df.info()` to examine column names, data types, and non-null values.
- `df.describe()` to obtain statistical summaries of numerical variables.
- `value_counts()` to analyze the frequency of categorical values.
- `isnull().sum()` to identify missing values.
- `groupby()` to calculate survival rates for different passenger groups.
- `crosstab()` to compare relationships between categorical variables.

These methods provided a basic understanding of the dataset before performing detailed visual analysis.

## Data Visualization

Data visualization was used to understand patterns and relationships in the Titanic dataset.

Different types of charts were created during the analysis, including:

- Count plots for categorical variables.
- Histograms for numerical distributions.
- Boxplots for identifying spread and potential outliers.
- Scatterplots for examining relationships between numerical variables.
- Bar plots for comparing survival rates.
- Correlation heatmaps for understanding relationships between numerical variables.
- Pairplots for multivariate analysis.

Each visualization was followed by observations to identify meaningful patterns and insights from the data.

## Conclusion

The Exploratory Data Analysis of the Titanic dataset revealed several important patterns and relationships.

The analysis showed that survival was strongly associated with gender and passenger class. Female passengers generally had higher survival rates than male passengers, while First Class passengers had better survival outcomes compared with Third Class passengers.

The analysis also identified missing values, variations in age and fare distributions, and potential outliers. Different visualizations helped in understanding the relationships between passenger characteristics and survival.

Overall, EDA provided useful insights into the Titanic dataset and demonstrated how statistical analysis and visualization can be used to discover meaningful patterns in real-world data.

```python
# import the required library
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load Titanic dataset
df = sns.load_dataset("titanic")

# Display first 5 rows
df.head()
```

```python
# Check the shape of the dataset
print("Number of rows and columns:", df.shape)

# Display column names
print("\nColumn Names:")
print(df.columns)

# Display dataset information
print("\nDataset Information:")
df.info()
```

```python
# Statistical summary of numerical columns
df.describe()
```

```python
# Check missing values in each column
missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)
```

```python
# Display only columns with missing values
print("Columns with Missing Values:")
print(missing_values[missing_values > 0])
```

### Observation:
The Titanic dataset contains missing values in the Age, Embarked, Deck, and Embark_town columns. The Deck column has the highest number of missing values, while Embarked and Embark_town have only a few missing values.

```python
# Count passengers by gender
gender_counts = df["sex"].value_counts()

print("Passenger Count by Gender:")
print(gender_counts)
```

### Observation:
The dataset contains both male and female passengers. The value_counts() function helps us understand the distribution of passengers by gender.

```python
# Count survived and non-survived passengers
survival_counts = df["survived"].value_counts()

print("Survival Count:")
print(survival_counts)
```

### Observation:
The survived column shows the number of passengers who survived and did not survive. A value of 0 represents passengers who did not survive, while 1 represents passengers who survived.

```python
# Rename sex column to gender
df.rename(columns={"sex": "gender"}, inplace=True)

# Check column names
print(df.columns)
```

```python
# Survival count by gender
survival_gender = pd.crosstab(df["gender"], df["survived"])

print("Survival by Gender:")
print(survival_gender)
```

```python
# Survival count by passenger class
survival_class = pd.crosstab(df["pclass"], df["survived"])

print("Survival by Passenger Class:")
print(survival_class)
```

```python
# Survival Count Visualization

sns.countplot(data=df, x="survived")

plt.title("Titanic Survival Count")
plt.xlabel("Survival Status")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation:
The bar chart shows the number of passengers who survived and did not survive. The number of passengers who did not survive is higher than the number of passengers who survived.

```python
# Survival by Gender Visualization

sns.countplot(data=df, x="gender", hue="survived")

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation:
The chart compares survival outcomes between male and female passengers. It shows that survival outcomes differed by gender, with female passengers having a higher survival count relative to their total number of passengers.

```python
# Survival by Passenger Class Visualization

sns.countplot(data=df, x="pclass", hue="survived")

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation:
The chart compares survival outcomes across the three passenger classes. It helps identify differences in survival between First, Second, and Third Class passengers.

📊 Chart meaning

1 = First Class

2 = Second Class

3 = Third Class

survived = 0 → Did not survive

survived = 1 → Survived

```python
# Age Distribution

plt.figure(figsize=(8, 5))

sns.histplot(data=df, x="age", bins=30, kde=True)

plt.title("Age Distribution of Titanic Passengers")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation:
The histogram shows the distribution of passenger ages in the Titanic dataset. Most passengers were concentrated in the young and middle-age groups, while relatively fewer passengers were present at very young and very old ages.

```python
# Age vs Fare Scatter Plot

plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="age", y="fare")

plt.title("Relationship Between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")

plt.show()
```

### Observation:
The scatter plot shows the relationship between passenger age and fare. The points are widely distributed, indicating that there is no strong clear relationship between age and fare. Some passengers paid unusually high fares compared with most passengers.

```python
# Age Boxplot

plt.figure(figsize=(8, 4))

sns.boxplot(data=df, x="age")

plt.title("Boxplot of Passenger Age")
plt.xlabel("Age")

plt.show()
```

### Observation:
The boxplot shows the distribution of passenger ages and helps identify the median, spread, and possible outliers. A few points may appear outside the whiskers, representing unusually high or low ages.

```python
# Fare Boxplot

plt.figure(figsize=(8, 4))

sns.boxplot(data=df, x="fare")

plt.title("Boxplot of Passenger Fare")
plt.xlabel("Fare")

plt.show()
```

### Observation:
The boxplot shows the distribution of passenger fares. Most passengers paid relatively lower fares, while a small number of passengers paid much higher fares, which appear as potential outliers.

```python
# Correlation Heatmap

plt.figure(figsize=(10, 7))

correlation = df.select_dtypes(include="number").corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap of Titanic Dataset")

plt.show()
```

### Observation:
The correlation heatmap shows the relationships between numerical variables in the Titanic dataset. It helps identify variables that have positive or negative relationships with each other and with passenger survival.

```python
# Pairplot of Titanic Dataset

sns.pairplot(
    df[["survived", "pclass", "age", "sibsp", "parch", "fare"]].dropna(),
    hue="survived"
)

plt.suptitle("Pairplot of Titanic Dataset", y=1.02)

plt.show()
```

### Observation:
The pairplot provides a visual overview of relationships among the numerical variables. It helps identify patterns, distributions, and possible relationships between passenger characteristics and survival.

```python
# Calculate survival rate by gender

gender_survival_rate = df.groupby("gender")["survived"].mean() * 100

print("Survival Rate by Gender:")
print(gender_survival_rate)
```

### Observation:
The survival rate differs between male and female passengers. The analysis helps identify the relationship between gender and the likelihood of survival.

```python
# Calculate survival rate by passenger class

class_survival_rate = df.groupby("pclass")["survived"].mean() * 100

print("Survival Rate by Passenger Class:")
print(class_survival_rate)
```

### Observation:
The survival rate varies across passenger classes. Passengers in higher classes generally had a higher survival rate than passengers in lower classes. This indicates that passenger class was an important factor associated with survival.

```python
# Survival Rate by Passenger Class

plt.figure(figsize=(8, 5))

sns.barplot(
    x=class_survival_rate.index,
    y=class_survival_rate.values
)

plt.title("Survival Rate by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate (%)")

plt.show()
```

### Observation:
The bar chart shows that survival rates were different across passenger classes. First Class passengers had the highest survival rate, while Third Class passengers had the lowest survival rate. This suggests that passenger class was strongly associated with survival.

```python
# Gender Distribution by Passenger Class

gender_class = pd.crosstab(df["pclass"], df["gender"])

print("Gender Distribution by Passenger Class:")
print(gender_class)
```

### Observation:
The table shows the distribution of male and female passengers across the three passenger classes. It helps us understand how gender was distributed among First, Second, and Third Class passengers.

```python
# Gender Distribution by Passenger Class Visualization

sns.countplot(data=df, x="pclass", hue="gender")

plt.title("Gender Distribution by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

```python
# Age Distribution by Gender

plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="age",
    hue="gender",
    bins=30,
    kde=True
)

plt.title("Age Distribution by Gender")
plt.xlabel("Age")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation:
The histogram compares the age distribution of male and female passengers. It shows how the ages of passengers were distributed across the two genders and helps identify the most common age groups.

```python
# Check for duplicate records

duplicate_count = df.duplicated().sum()

print("Number of Duplicate Rows:", duplicate_count)
```

### Observation:
The duplicate check helps identify whether the dataset contains repeated records. Duplicate records can affect the accuracy of exploratory analysis and should be considered before further analysis.

```python
# Check data types of all columns

print("Data Types of Each Column:")
print(df.dtypes)
```

### Observation:
The dataset contains both numerical and categorical variables. Numerical columns include variables such as age, fare, pclass, and survived, while categorical columns include gender, embarked, class, and other descriptive attributes.

```python
# Gender Distribution

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="gender")

plt.title("Gender Distribution of Titanic Passengers")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation:
The bar chart shows the distribution of male and female passengers in the Titanic dataset. It helps us understand the number of passengers belonging to each gender category.

```python
# Passenger Class Distribution

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="pclass")

plt.title("Passenger Class Distribution")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation:
The bar chart shows the distribution of passengers across the three passenger classes. It helps identify which passenger class had the highest and lowest number of passengers.

```python
# Embarkation Port Distribution

plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="embarked")

plt.title("Passenger Distribution by Embarkation Port")
plt.xlabel("Embarkation Port")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation:
The bar chart shows the number of passengers who embarked from each port. It helps identify which embarkation port had the highest and lowest number of passengers.

```python
# Survival Rate by Gender

gender_survival_rate = df.groupby("gender")["survived"].mean() * 100

plt.figure(figsize=(7, 5))

sns.barplot(
    x=gender_survival_rate.index,
    y=gender_survival_rate.values
)

plt.title("Survival Rate by Gender")
plt.xlabel("Gender")
plt.ylabel("Survival Rate (%)")

plt.show()
```

### Observation:
The bar chart shows the survival rate of passengers by gender. Female passengers had a higher survival rate than male passengers, indicating a strong difference in survival outcomes between the two gender groups.

```python
# Create age groups

df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 12, 18, 35, 60, 100],
    labels=["Child", "Teenager", "Young Adult", "Adult", "Senior"]
)

# Display age groups
print(df["age_group"].value_counts())
```

### Observation:
The age column has been divided into meaningful age groups: Child, Teenager, Young Adult, Adult, and Senior. This grouping makes it easier to compare survival patterns across different age categories.

```python
# Survival Rate by Age Group

age_survival_rate = df.groupby("age_group", observed=True)["survived"].mean() * 100

print("Survival Rate by Age Group:")
print(age_survival_rate)
```

### Observation:
The survival rate varies across different age groups. Comparing these groups helps identify whether age was associated with different survival outcomes among Titanic passengers.

```python
# Visualize Survival Rate by Age Group

plt.figure(figsize=(9, 5))

sns.barplot(
    x=age_survival_rate.index,
    y=age_survival_rate.values
)

plt.title("Survival Rate by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Survival Rate (%)")
plt.xticks(rotation=20)

plt.show()
```

### Observation:
The bar chart compares survival rates across the different age groups. It provides a clear visual comparison of survival outcomes among children, teenagers, young adults, adults, and seniors.

```python
# Multivariate Analysis: Gender, Passenger Class and Survival

plt.figure(figsize=(9, 6))

sns.countplot(
    data=df,
    x="pclass",
    hue="gender",
    palette="Set2"
)

plt.title("Gender Distribution Across Passenger Classes")
plt.xlabel("Passenger Class")
plt.ylabel("Number of Passengers")

plt.show()
```

### Observation:
The chart compares gender distribution across different passenger classes. It helps identify how male and female passengers were distributed among First, Second, and Third Class.

```python
# Survival by Passenger Class and Gender

sns.catplot(
    data=df,
    x="pclass",
    hue="gender",
    col="survived",
    kind="count",
    height=5
)

plt.show()
```

### Observation:
The multivariate analysis shows how passenger class, gender, and survival are related. It provides a combined view of survival patterns and helps identify differences between male and female passengers across different passenger classes.
